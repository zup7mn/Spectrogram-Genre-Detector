This notebook is for converting the .ogg files into the modified spectrogram images, and then saving those in my DataBase directory. This code is (idealy) used only once in the scope of my project.

```python
from dataprocessing_utils import spectrogram_of_middle, save_spec_array_to_file
import os
```

```python
def save_all_songs_in_directory(output_db, input_db="MusicDB", labels_file_name="labels.csv"):
    """
    Saves all songs in the input_db directory as spectrograms in the output_db directory.
    The labels are saved in a CSV file.
    
    :param output_db: Directory where the spectrograms will be saved.
    :param input_db: Directory containing the music files.
    :param labels_file_name: Name of the CSV file to save labels.
    """
    if not os.path.exists(output_db):
        os.makedirs(output_db)

    for genre in os.listdir(input_db):
        genre_path = os.path.join(input_db, genre)
        if os.path.isdir(genre_path):
            for song in os.listdir(genre_path):
                song_path = os.path.join(genre_path, song)
                if song.endswith('.ogg'):
                    spec = spectrogram_of_middle(song_path)
                    save_spec_array_to_file(spec, genre, db=output_db, labels_file_name=labels_file_name)
```
